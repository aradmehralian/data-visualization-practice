```python
g = sns.histplot(data=df, x="petal_length", hue="species", multiple="layer")
_ = g.set_title(
    "Histogram of Petal Lengths, by Species", fontweight="bold", fontsize=11
)
```

```python
plt.figure(figsize=(7, 5))
sns.kdeplot(data=df, x="petal_length", hue="species", fill=True)
plt.title("Distribution of Petal Lengths, by Species", fontweight="bold", fontsize=12)
plt.tight_layout()
```

```python
# 3
malignant_area_mean = df[df["Diagnosis"] == "M"]["Area (mean)"].mean()
benign_area_mean = df[df["Diagnosis"] == "B"]["Area (mean)"].mean()
```

```python
g = sns.histplot(data=df, x="Area (mean)", hue="Diagnosis")
g.axvline(
    x=benign_area_mean,
    color="green",
    linestyle="--",
    label="Benign Tumor Mean Area",
)
g.axvline(
    x=malignant_area_mean,
    color="red",
    linestyle="--",
    label="Malignant Tumor Mean Area",
)
g.annotate(text=f"Mean:\n{malignant_area_mean:.2f}", xy=(malignant_area_mean + 10, 90))
g.annotate(text=f"Mean:\n{benign_area_mean:.2f}", xy=(benign_area_mean + 10, 90))

g.set_title("Histogram of Tumor Area, by Diagnosis", title_params)
g.legend()

g.figure.tight_layout()
```

```python
# 4
g = sns.kdeplot(data=df, x="Radius (worst)", hue="Diagnosis", fill=True)
g.set_title("Distribution of the Radius of Tumors, by Diagnosis", title_params)
g.legend(labels=["Malignant", "Benign"])
plt.tight_layout()
```
