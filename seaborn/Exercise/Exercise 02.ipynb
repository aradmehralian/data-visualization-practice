```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib.dates as mdates
```

```python
sns.set_theme(
    context="notebook", style="white", rc={"figure.figsize": (14, 6)}, palette="tab10"
)
```

```python
df = pd.read_csv("data/spotify.csv", index_col="Date", parse_dates=True)
```

```python
df.head(10)
```

```python
df.tail()
```

```python
fig, ax = plt.subplots(figsize=(12, 6))

sns.lineplot(data=df, ax=ax)
ax.set_title("Daily Global Streams of Popular Songs in 2017-2018", fontweight="bold")

fig.tight_layout()
```

```python
list(df.columns)
```

```python
ax = sns.lineplot(data=df, x="Date", y="Shape of You", label="Shape of You")
sns.lineplot(data=df, ax=ax, x="Date", y="Despacito", label="Despacito")
ax.set(ylabel=None)
ax.legend(title="Name of Song")

plt.tight_layout()
```

```python
print(df.loc["2018-07-01", "Chinese American Museum"])
```

```python
avila_oct2018 = df.loc["2018-10-01", "Avila Adobe"]
firehouese_oct2018 = df.loc["2018-10-01", "Firehouse Museum"]
print(avila_oct2018 - firehouese_oct2018)
```

```python
# 3
fig, ax = plt.subplots(figsize=(12, 6))
sns.lineplot(data=df, ax=ax)
ax.set(ylabel="Number of Visitors")
ax.set_title("Monthly Visitors to Los Angeles Museums", fontweight="bold")

fig.tight_layout()
```

```python
# 4
ax = sns.lineplot(data=df["Avila Adobe"], color="navy")
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y - %b"))
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=6))

for label in ax.get_xticklabels():
    label.set_rotation(45)

ax.figure.tight_layout()
```
