```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("data/insurance.csv")
```

```python
df
```

```python
len(df)
```

```python
df.info()
```

```python
ax = sns.scatterplot(data=df, x="bmi", y="charges")
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
sns.regplot(data=df, x="bmi", y="charges", color="navy")

fig.tight_layout()
```

```python
ax = sns.scatterplot(data=df, x="bmi", y="charges", hue="smoker")
```

```python
ax = sns.lmplot(
    data=df, x="bmi", y="charges", hue="smoker", height=5, aspect=1.5
)  # figure-level plot
```

```python
type(ax)
```

```python
plt.figure(figsize=(10, 7))
sns.swarmplot(data=df, x="smoker", y="charges", size=4, hue="sex")
plt.title("Distribution of Insurance Fee based on Smoking Status", fontweight="bold")
plt.tight_layout()
```
