# Seaborn Exercise 3

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("data/flight_delays.csv", index_col="Month")
```

```python
df.head()
```

```python
df.index = pd.to_datetime(df.index, format="%m").strftime("%B")
```

```python
df
```

```python
fig, ax = plt.subplots(figsize=(12, 6))
title_params = dict(fontweight="bold", fontsize=12)

sns.barplot(data=df, ax=ax, x="Month", y="NK", palette="Set3", hue="Month")
ax.set_title("Average Arrival Delay for Spirit Airlines' Flights", title_params)
ax.set_ylabel("Flight Delay (minutes)")

fig.tight_layout()
```

```python
ax = sns.heatmap(data=df, cmap="coolwarm", annot=True)
ax.set_xlabel("Airline")
ax.set_title("Average Arrival Delay for Each Airline, by Month", title_params)

ax.figure.set_size_inches((9, 7))
ax.figure.tight_layout()
```
